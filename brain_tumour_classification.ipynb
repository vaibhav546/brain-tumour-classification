# Unzipping a File with Python

This script demonstrates how to extract the contents of a `.zip` file using Python's built-in `zipfile` module.

## Code Explanation

Ensure that the zipfile module is imported before attempting to unzip files.
The extractall() method extracts all files in the ZIP archive. If you want to extract specific files, consider using the extract() method instead.

```python
# Import the necessary module
import zipfile

# Define the path to the zip file
zip_file_path = "archive.zip"

# Define the directory to extract to
extract_dir = "./"

# Open the zip file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    # Extract all the contents into the specified directory
    zip_ref.extractall(extract_dir)

print("Extraction completed!")
```

```python
import os
import shutil
import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import VGG19
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Flatten, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.optimizers.legacy import SGD  # Updated import
from sklearn.utils import shuffle
import imutils
import time

# Function to rename files in a directory

def rename_files(folder, prefix):
    """
    Renames all files in a given directory with a specified prefix followed by a number.

    Parameters:
    folder (str): The directory containing the files to rename.
    prefix (str): The prefix to use for the renamed files.

    """
    count = 1
    for filename in os.listdir(folder):
        source = os.path.join(folder, filename)
        destination = os.path.join(folder, f"{prefix}_{count}.jpg")
        os.rename(source, destination)
        count += 1
    print(f"All files are renamed in the {prefix} dir.")

# Function to count files in a directory
def count_files(folder):
    return len(os.listdir(folder))

# Function to plot bar chart
def plot_bar_chart(data, title, xlabel, ylabel):
    """
    Plots a bar chart from a given dictionary.

    Parameters:
    data (dict): The dictionary containing the data to plot.
    title (str): The title of the plot.
    xlabel (str): The label for the x-axis.
    ylabel (str): The label for the y-axis.
    """
    typex = data.keys()
    values = data.values()
    plt.figure(figsize=(5, 7))
    plt.bar(typex, values, color="red")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.show()

# Function to augment data
def augmented_data(file_dir, n_generated_samples, save_to_dir):
    """
    Generates augmented images from a given directory and saves them to another directory.

    Parameters:
    file_dir (str): The directory containing the original images.
    n_generated_samples (int): The number of augmented images to generate per original image.
    save_to_dir (str): The directory to save the augmented images.
    """
    data_gen = ImageDataGenerator(rotation_range=10,
                                  width_shift_range=0.1,
                                  height_shift_range=0.1,
                                  shear_range=0.1,
                                  brightness_range=(0.3, 1.0),
                                  horizontal_flip=True,
                                  vertical_flip=True,
                                  fill_mode='nearest')
    for filename in os.listdir(file_dir):
        image = cv2.imread(os.path.join(file_dir, filename))
        image = image.reshape((1,) + image.shape)
        save_prefix = 'aug_' + filename[:-4]
        i = 0
        for batch in data_gen.flow(x=image, batch_size=1, save_to_dir=save_to_dir, save_prefix=save_prefix, save_format="jpg"):
            i += 1
            if i >= n_generated_samples:
                break

# Function to get time in hh:mm:ss format
def timing(sec_elapsed):
    h = int(sec_elapsed / (60 * 60))
    m = int(sec_elapsed % (60 * 60) / 60)
    s = sec_elapsed % 60
    return f"{h}:{m}:{s}"

# Function to summarize data
def data_summary(main_path):
    """
    Summarizes the number of positive and negative samples in the dataset.

    Parameters:
    main_path (str): The main directory path containing 'yes' and 'no' subdirectories.
    """
    yes_path = os.path.join(main_path, "yes/")
    no_path = os.path.join(main_path, "no/")
    n_pos = count_files(yes_path)
    n_neg = count_files(no_path)
    n = n_pos + n_neg
    pos_per = (n_pos * 100) / n
    neg_per = (n_neg * 100) / n
    print(f"Number of samples: {n}")
    print(f"Number of positive samples: {n_pos} ({pos_per:.2f}%)")
    print(f"Number of negative samples: {n_neg} ({neg_per:.2f}%)")

# Function to crop brain tumor from image
def crop_brain_tumor(image, plot=False):
    """
    Crops the brain tumor from an image and optionally plots the original and cropped images.

    Parameters:
    image (ndarray): The input image containing a brain scan.
    plot (bool): Whether to plot the original and cropped images. Default is False.

    Returns:
    ndarray: The cropped image containing the brain tumor.
    """
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (5, 5), 0)
    thres = cv2.threshold(gray, 45, 255, cv2.THRESH_BINARY)[1]
    thres = cv2.erode(thres, None, iterations=2)
    thres = cv2.dilate(thres, None, iterations=2)
    cnts = cv2.findContours(thres.copy(), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    cnts = imutils.grab_contours(cnts)
    c = max(cnts, key=cv2.contourArea)
    extLeft = tuple(c[c[:, :, 0].argmin()][0])
    extRight = tuple(c[c[:, :, 0].argmax()][0])
    extTop = tuple(c[c[:, :, 1].argmin()][0])
    extBot = tuple(c[c[:, :, 1].argmax()][0])
    new_image = image[extTop[1]:extBot[1], extLeft[0]:extRight[0]]
    if plot:
        plt.figure()
        plt.subplot(1, 2, 1)
        plt.imshow(image)
        plt.tick_params(axis='both', which='both', top=False, bottom=False, left=False, right=False, labelbottom=False, labeltop=False, labelleft=False, labelright=False)
        plt.title('Original Image')
        plt.subplot(1, 2, 2)
        plt.imshow(new_image)
        plt.tick_params(axis='both', which='both', top=False, bottom=False, left=False, right=False, labelbottom=False, labeltop=False, labelleft=False, labelright=False)
        plt.title('Cropped Image')
        plt.show()
    return new_image

# Function to process and save cropped images
def process_and_save_cropped_images(folder):
    """
    Processes and saves cropped brain tumor images in the specified folder.

    Parameters:
    folder (str): The directory containing the images to process.
    """
    for filename in os.listdir(folder):
        img = cv2.imread(os.path.join(folder, filename))
        img = crop_brain_tumor(img, False)
        cv2.imwrite(os.path.join(folder, filename), img)

# Function to load data
def load_data(dir_list, image_size):
    """
    Loads and processes images from the specified directories, resizes them, and normalizes the pixel values.

    Parameters:
    dir_list (list): List of directories containing the images.
    image_size (tuple): The desired image size (width, height).

    Returns:
    tuple: A tuple containing the processed images (X) and their corresponding labels (y).
    """
    X = []
    y = []
    image_width, image_height = image_size
    for directory in dir_list:
        for filename in os.listdir(directory):
            image = cv2.imread(os.path.join(directory, filename))
            image = crop_brain_tumor(image, plot=False)
            image = cv2.resize(image, (image_width, image_height), interpolation=cv2.INTER_CUBIC)
            image = image / 255.0
            X.append(image)
            y.append(1 if directory.endswith("yes") else 0)
    X = np.array(X)
    y = np.array(y)
    X, y = shuffle(X, y)
    print(f"Number of examples: {len(X)}")
    print(f"X shape: {X.shape}")
    print(f"y shape: {y.shape}")
    return X, y

# Function to plot sample images
def plot_sample_images(X, y, n=50):
    for label in [0, 1]:
        images = X[np.argwhere(y == label)]
        n_images = images[:n]
        columns_n = 10
        rows_n = n // columns_n
        plt.figure(figsize=(20, 10))
        i = 1
        for image in n_images:
            plt.subplot(rows_n, columns_n, i)
            plt.imshow(image[0])
            plt.tick_params(axis='both', which='both', top=False, bottom=False, left=False, right=False, labelbottom=False, labeltop=False, labelleft=False, labelright=False)
            i += 1
        label_to_str = lambda label: "Yes" if label == 1 else "No"
        plt.suptitle(f"Brain Tumor: {label_to_str(label)}")
        plt.show()

# Function to split data into training, testing, and validation sets
def split_data(base_dir, augmented_yes_path, augmented_no_path):
    """
    Splits the data into training, testing, and validation sets and creates directories for each set.

    Parameters:
    base_dir (str): The base directory where the train, test, and valid directories will be created.
    augmented_yes_path (str): Path to the directory containing augmented 'yes' images.
    augmented_no_path (str): Path to the directory containing augmented 'no' images.
    """
    dirs = {
        "train_tumorous": os.path.join(base_dir, 'train/tumorous'),
        "test_tumorous": os.path.join(base_dir, 'test/tumorous'),
        "valid_tumorous": os.path.join(base_dir, 'valid/tumorous'),
        "train_nontumorous": os.path.join(base_dir, 'train/nontumorous'),
        "test_nontumorous": os.path.join(base_dir, 'test/nontumorous'),
        "valid_nontumorous": os.path.join(base_dir, 'valid/nontumorous'),
    }

    for dir_path in dirs.values():
        os.makedirs(dir_path, exist_ok=True)

    split_files(augmented_yes_path, dirs["train_tumorous"], dirs["test_tumorous"], dirs["valid_tumorous"], [759, 922])
    split_files(augmented_no_path, dirs["train_nontumorous"], dirs["test_nontumorous"], dirs["valid_nontumorous"], [686, 833])

def split_files(src_dir, train_dir, test_dir, valid_dir, split_points):
    """
    Splits files from a source directory into training, testing, and validation sets based on split points.

    Parameters:
    src_dir (str): Source directory containing the files to split.
    train_dir (str): Destination directory for training files.
    test_dir (str): Destination directory for testing files.
    valid_dir (str): Destination directory for validation files.
    split_points (list): Indices indicating where to split the files for train, test, and validation sets.
    """
    files = os.listdir(src_dir)
    train_files, test_files, valid_files = files[:split_points[0]], files[split_points[0]:split_points[1]], files[split_points[1]:]

    copy_files(src_dir, train_dir, train_files)
    copy_files(src_dir, test_dir, test_files)
    copy_files(src_dir, valid_dir, valid_files)

def copy_files(src_dir, dst_dir, files):
    """
    Copies files from a source directory to a destination directory.

    Parameters:
    src_dir (str): Source directory containing the files to copy.
    dst_dir (str): Destination directory where the files will be copied.
    files (list): List of filenames to copy.
    """
    for fname in files:
        src = os.path.join(src_dir, fname)
        dst = os.path.join(dst_dir, fname)
        shutil.copyfile(src, dst)

# Function to create data generators
def create_generators(base_dir, image_size=(224, 224), batch_size=32):
    """
    Creates data generators for training, testing, and validation sets.

    Parameters:
    base_dir (str): The base directory containing the train, test, and valid directories.
    image_size (tuple): The target size of the images (width, height). Default is (224, 224).
    batch_size (int): The size of the batches. Default is 32.

    Returns:
    tuple: A tuple containing the train, test, and validation data generators.
    """
    datagen = ImageDataGenerator(rescale=1.0/255,
                                 rotation_range=20,
                                 zoom_range=0.2,
                                 shear_range=0.2,
                                 width_shift_range=0.4,
                                 height_shift_range=0.4,
                                 fill_mode='nearest')

    train_generator = datagen.flow_from_directory(
        os.path.join(base_dir, 'train/'),
        target_size=image_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=True
    )

    valid_generator = datagen.flow_from_directory(
        os.path.join(base_dir, 'valid/'),
        target_size=image_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=False
    )

    test_generator = datagen.flow_from_directory(
        os.path.join(base_dir, 'test/'),
        target_size=image_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=False
    )

    return train_generator, valid_generator, test_generator

from tensorflow.keras.applications import VGG19
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Flatten, Dense, Dropout, BatchNormalization, Input
from tensorflow.keras.optimizers import SGD, Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
import matplotlib.pyplot as plt
import os

# Function to build the model with additional improvements
def build_model(input_shape, base_model_weights='imagenet', trainable_layers=[]):
    """
    Builds and compiles a VGG19-based model with additional dense layers and dropout for classification.

    Parameters:
    input_shape (tuple): Shape of the input images (height, width, channels).
    base_model_weights (str): Weights to load into the base model. Default is 'imagenet'.
    trainable_layers (list): List of layer names in the base model to set as trainable.

    Returns:
    Model: The compiled model.
    """
    base_model = VGG19(input_shape=input_shape, include_top=False, weights=base_model_weights)

    for layer in base_model.layers:
        layer.trainable = False

    x = base_model.output
    flat = Flatten()(x)
    class_1 = Dense(4608, activation='relu')(flat)
    bn_1 = BatchNormalization()(class_1)
    drop_out_1 = Dropout(0.5)(bn_1)
    class_2 = Dense(2304, activation='relu')(drop_out_1)
    bn_2 = BatchNormalization()(class_2)
    drop_out_2 = Dropout(0.5)(bn_2)
    class_3 = Dense(1152, activation='relu')(drop_out_2)
    bn_3 = BatchNormalization()(class_3)
    drop_out_3 = Dropout(0.5)(bn_3)
    output = Dense(2, activation='softmax')(drop_out_3)

    model = Model(base_model.input, output)

    for layer in base_model.layers:
        if layer.name in trainable_layers:
            layer.trainable = True

    optimizer = Adam(learning_rate=0.0001)
    model.compile(loss='categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])

    return model

# Function to train the model with increased epochs and modified callbacks
def train_model(model, train_generator, valid_generator, epochs=50, checkpoint_path='model.keras'):
    """
    Trains the model using the training and validation generators, with early stopping and learning rate reduction.

    Parameters:
    model (Model): The compiled model to train.
    train_generator (DirectoryIterator): Generator for the training data.
    valid_generator (DirectoryIterator): Generator for the validation data.
    epochs (int): Number of epochs to train. Default is 50.
    checkpoint_path (str): Path to save the best model checkpoint.

    Returns:
    History: Training history object.
    """
    es = EarlyStopping(monitor='val_loss', verbose=1, mode='min', patience=5)
    cp = ModelCheckpoint(checkpoint_path, monitor='val_loss', verbose=1, save_best_only=True, save_weights_only=False, mode='auto', save_freq='epoch')
    lrr = ReduceLROnPlateau(monitor='val_loss', patience=3, verbose=1, factor=0.5, min_lr=1e-6)

    steps_per_epoch = train_generator.samples // train_generator.batch_size
    validation_steps = valid_generator.samples // valid_generator.batch_size

    history = model.fit(
        train_generator,
        steps_per_epoch=steps_per_epoch,
        epochs=epochs,
        callbacks=[es, cp, lrr],
        validation_data=valid_generator,
        validation_steps=validation_steps
    )

    return history

# Function to plot training performance
def plot_training_performance(history, filename='training_performance.jpeg'):
    """
    Plots the training and validation accuracy and loss over epochs.

    Parameters:
    history (History): Training history object.
    filename (str): Filename to save the plot. Default is 'training_performance.jpeg'.
    """
    fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(12, 6))
    fig.suptitle("Model Training", fontsize=12)
    epochs_list = list(range(1, len(history.history['accuracy']) + 1))

    ax1.plot(epochs_list, history.history['accuracy'], color='b', linestyle='-', label='Training Data')
    ax1.plot(epochs_list, history.history['val_accuracy'], color='r', linestyle='-', label='Validation Data')
    ax1.set_title('Training Accuracy', fontsize=12)
    ax1.set_xlabel('Epochs', fontsize=12)
    ax1.set_ylabel('Accuracy', fontsize=12)
    ax1.legend(frameon=False, loc='lower center', ncol=2)

    ax2.plot(epochs_list, history.history['loss'], color='b', linestyle='-', label='Training Data')
    ax2.plot(epochs_list, history.history['val_loss'], color='r', linestyle='-', label='Validation Data')
    ax2.set_title('Training Loss', fontsize=12)
    ax2.set_xlabel('Epochs', fontsize=12)
    ax2.set_ylabel('Loss', fontsize=12)
    ax2.legend(frameon=False, loc='upper center', ncol=2)

    plt.savefig(filename, format='jpeg', dpi=100, bbox_inches='tight')
    plt.show()

# Main script
def main():
    dataset_dir = '/content/brain_tumor_dataset'
    augmented_dir = 'augmented_data'
    base_dir = 'tumorous_and_nontumorous'

    yes_path = os.path.join(dataset_dir, 'yes')
    no_path = os.path.join(dataset_dir, 'no')

    # Print initial number of images in each category
    n_yes = count_files(yes_path)
    n_no = count_files(no_path)
    print(f"Number of tumor images: {n_yes}")
    print(f"Number of non-tumor images: {n_no}")
    print(f"Total images: {n_yes + n_no}")

    # Plot bar graph
    data = {'Tumor': n_yes, 'Non-Tumor': n_no}
    plot_bar_chart(data, 'Distribution of Tumor and Non-Tumor Images', 'Category', 'Number of Images')

    rename_files(yes_path, 'yes')
    rename_files(no_path, 'no')

    # Load and plot sample images
    image_size = (240, 240)
    X, y = load_data([yes_path, no_path], image_size)
    plot_sample_images(X, y, n=20)  # Plot 20 sample images for each category

    if not os.path.exists(augmented_dir):
        os.makedirs(augmented_dir)

    augmented_yes_path = os.path.join(augmented_dir, 'yes')
    augmented_no_path = os.path.join(augmented_dir, 'no')

    os.makedirs(augmented_yes_path, exist_ok=True)
    os.makedirs(augmented_no_path, exist_ok=True)

    augmented_data(yes_path, 6, augmented_yes_path)
    augmented_data(no_path, 9, augmented_no_path)

    split_data(base_dir, augmented_yes_path, augmented_no_path)

    train_generator, test_generator, valid_generator = create_generators(base_dir, image_size=(240, 240))

    model_01 = build_model(input_shape=(240, 240, 3))
    history_01 = train_model(model_01, train_generator, valid_generator)
    plot_training_performance(history_01, 'training_frozencnn.jpeg')

    if not os.path.isdir('saved_models/'):
        os.mkdir('saved_models/')
    model_01.save(filepath="saved_models/vgg19_model_01.h5", overwrite=True)

    if not os.path.isdir('model_weights/'):
        os.mkdir('model_weights/')
    model_01.save_weights(filepath="model_weights/vgg19_model_01.weights.h5", overwrite=True)

    model_01.load_weights("model_weights/vgg19_model_01.weights.h5")
    vgg_val_eval_01 = model_01.evaluate(valid_generator)
    vgg_test_eval_01 = model_01.evaluate(test_generator)

    print(f'Validation Loss: {vgg_val_eval_01[0]}')
    print(f'Validation Acc: {vgg_val_eval_01[1]}')
    print(f'Testing Loss: {vgg_test_eval_01[0]}')
    print(f'Testing Acc: {vgg_test_eval_01[1]}')

    model_02 = build_model(input_shape=(240, 240, 3), trainable_layers=['block5_conv4', 'block5_conv3'])
    model_02.load_weights('model_weights/vgg19_model_01.weights.h5')
    history_02 = train_model(model_02, train_generator, valid_generator)
    plot_training_performance(history_02, 'training_finetunedcnn.jpeg')

if __name__ == '__main__':
    main()

```

# Brain Tumor Detection Flask Application

Flask: A micro web framework for Python to handle web requests and responses.
pyngrok: A package to set up an ngrok tunnel, making the local Flask server accessible over the internet.
tensorflow: A deep learning framework used to load the pre-trained model and perform predictions.

1. Application Structure
 Importing Required Libraries
The script starts by importing the necessary libraries for Flask, TensorFlow, and ngrok.

2. def model_predict(img_path, model):
Prediction Function
This function preprocesses the uploaded image and uses the model to make a prediction.

3. Home Route
The home route displays a simple HTML form where users can upload an image for prediction.

4. The predict route handles the uploaded image, runs it through the model, and returns the prediction as JSON.

5. To make the local Flask server accessible over the internet, the application sets up an ngrok tunnel. The Flask app runs on port 5000.


```python
from flask import Flask, request, render_template_string, jsonify
from pyngrok import ngrok
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
import numpy as np
import os

app = Flask(__name__)

# Load your trained model
model = load_model('/content/saved_models/vgg19_model_01.h5')

def model_predict(img_path, model):
    img = image.load_img(img_path, target_size=(240, 240))  # Adjust target size as per your model
    x = image.img_to_array(img)
    x = np.expand_dims(x, axis=0)
    x = x / 255.0  # Scale image pixels if necessary
    preds = model.predict(x)
    return preds

@app.route('/', methods=['GET'])
def index():
    return '''
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Brain Tumor Detection</title>
    </head>
    <body>
        <h1>Brain Tumor Detection</h1>
        <form action="/predict" method="post" enctype="multipart/form-data">
            <input type="file" name="file" accept="image/*">
            <button type="submit" class="btn ">Predict</button>
        </form>
        <div id="result"></div>
        <script>
            document.querySelector('form').onsubmit = async function(event) {
                event.preventDefault();
                const formData = new FormData(event.target);
                const response = await fetch('/predict', {
                    method: 'POST',
                    body: formData
                });
                const result = await response.json();
                document.getElementById('result').textContent = 'Prediction: ' + result.prediction;
            };
        </script>
    </body>
    </html>
    '''

@app.route('/predict', methods=['POST'])
def upload():
    if 'file' not in request.files:
        return jsonify({"error": "No file part"})
    file = request.files['file']
    if file.filename == '':
        return jsonify({"error": "No selected file"})
    if file:
        file_path = os.path.join('uploads', file.filename)
        if not os.path.exists('uploads'):
            os.makedirs('uploads')
        file.save(file_path)
        preds = model_predict(file_path, model)
        result = np.argmax(preds, axis=1)
        return jsonify({"prediction": "Yes" if result[0] == 1 else "No"})
    return jsonify({"error": "Failed to process the file"})

if __name__ == '__main__':
    # Set up ngrok
    public_url = ngrok.connect(5000)
    print(f" * ngrok tunnel \"{public_url}\" -> \"http://127.0.0.1:5000\"")

    # Start Flask app
    app.run()
```
